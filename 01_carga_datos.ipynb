```python
# Cargar el dataset de ejemplo incluido en Databricks
file_path = "dbfs:/databricks-datasets/wine-quality/winequality-red.csv"

df = spark.read.csv(file_path, header=True, inferSchema=True, sep=";")

# Limpiar nombres de columna (vienen con espacios)
for col in df.columns:
    df = df.withColumnRenamed(col, col.strip().replace(" ", "_"))

display(df)
print(f"Filas cargadas: {df.count()}, Columnas: {len(df.columns)}")

# Persistir como tabla Delta
df.write.format("delta").mode("overwrite").saveAsTable("wine_quality_raw")

print("Tabla Delta 'wine_quality_raw' creada correctamente.")
```
